```python
# Lists the matches for a specific competition and season, in this case UEFA Euro 2024
from statsbombpy import sb

matches = sb.matches(competition_id=55, season_id=282)
matches[["match_id", "match_date", "home_team", "away_team", "home_score", "away_score"]].sort_values("match_date")
```

```python
# Pulls out the final match from the UEFA Euro 2024 competition and season
from statsbombpy import sb

# Finds the final match ID
final = matches.sort_values("match_date").iloc[-1]
match_id = final["match_id"]

# Prints the result for the final match
print(final["home_team"], final["home_score"], "-", final["away_score"], final["away_team"], "| match_id:", match_id)
```

```python
# Loads every event from the Euro 2024 final
from statsbombpy import sb

# Finds the final match ID
final = matches.sort_values("match_date").iloc[-1]
match_id = final["match_id"]

# Loads all the events from the match
events = sb.events(match_id=match_id)
print(events.shape)
events.head()

# Saves the events locally
events.to_pickle("euro2024_final_events.pkl")
```

```python
# Shows the top five carriers of the ball in the final
from statsbombpy import sb
import pandas as pd

events = pd.read_pickle("euro2024_final_events.pkl")

carries = events[events["type"] == "Carry"]
carries.groupby(["team", "player"]).agg(
    carries=("duration", "count"),
    total_seconds=("duration", "sum"),
).sort_values("carries", ascending=False).head(5).round(1)
```

```python
# Challenge 1: Who was the most dangerous? For every play who took a shot in the final, how many shots did they take and what was their total xG?
import pandas as pd

events = pd.read_pickle("euro2024_final_events.pkl")

shots = events[events["type"] == "Shot"]
shots.groupby(["team", "player"]).agg(
    shots=("shot_statsbomb_xg", "count"),
    total_xg=("shot_statsbomb_xg", "sum"),
    goals_scored=("shot_outcome", lambda x: (x == "Goal").sum())
).sort_values("total_xg", ascending=False).round(2)
```

```python
# Challenge 2: Who kept the ball better? What percentage of each team's passes were completed?

import pandas as pd

events = pd.read_pickle("euro2024_final_events.pkl")

# All passes
passes = events[events["type"] == "Pass"]

# Only passes which are NaN (aka don't have a type assigned to them which would mean they're not from open play)
open_play_passes = passes[passes["pass_type"].isna()]

# Groups the open play passes by team and turns them into a percentage and rounds to 1 dp, and sorts them in descending order
(open_play_passes.groupby("team")["pass_outcome"].apply(lambda col: col.isna().mean()) * 100).round(1).sort_values(ascending=False)
```
